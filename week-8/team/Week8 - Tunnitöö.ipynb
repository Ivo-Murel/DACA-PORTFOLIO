```python
# 2. samm: Müügiandmete pärimise funktsioon kuupäevafiltrite ja veakäsitlusega
from supabase import create_client
import os
import pandas as pd
from dotenv import load_dotenv

load_dotenv()
url = os.getenv("SUPABASE_URL")
key = os.getenv("SUPABASE_KEY")
supabase = create_client(url, key)

def fetch_sales(start_date=None, end_date=None):
    """Pärib Supabase'ist müügiandmed valikuliste kuupäevafiltritega."""
    try:
        query = supabase.table("sales").select("*")
        
        # Kui kuupäevad on ette antud, rakendame filtrid
        if start_date:
            query = query.gte("sale_date", start_date)
        if end_date:
            query = query.lte("sale_date", end_date)
            
        response = query.execute()
        df_sales = pd.DataFrame(response.data)
        return df_sales
    except Exception as e:
        print(f"[VIGA] Müügiandmete pärimine ebaõnnestus: {e}")
        return pd.DataFrame()

print("Funktsioon fetch_sales() on edukalt loodud!")
```

```python
# 3. samm: Kliendi- ja tooteandmete pärimise funktsioonid
def fetch_customers():
    """Pärib Supabase'ist kliendiandmed."""
    try:
        response = supabase.table("customers").select("*").execute()
        return pd.DataFrame(response.data)
    except Exception as e:
        print(f"[VIGA] Kliendiandmete pärimine ebaõnnestus: {e}")
        return pd.DataFrame()

def fetch_products():
    """Pärib Supabase'ist tooteandmed."""
    try:
        response = supabase.table("products").select("*").execute()
        return pd.DataFrame(response.data)
    except Exception as e:
        print(f"[VIGA] Tooteandmete pärimine ebaõnnestus: {e}")
        return pd.DataFrame()

print("Funktsioonid fetch_customers() ja fetch_products() on loodud!")
```

```python
# Supabase'i API tagastab vaikimisi korraga maksimaalselt 1000 rida. Suurte andmemahtude jaoks teeme tsükli, mis laeb andmeid lehtede kaupa (näiteks 1000 rida korraga), kuni rohkem ridu pole jäänud. Lisaks lisame time ja try-except abil korduskatse loogika, kui võrgus peaks esinema tõrge.
# Edasijõudnute tase: Funktsioonid paginatsiooni ja exponential backoff retry loogikaga
import time
import pandas as pd

def fetch_with_pagination_and_retry(table_name: str, batch_size: int = 1000, max_retries: int = 3) -> pd.DataFrame:
    """
    Pärib Supabase tabelist kõik read paginatsiooni abil, 
    kasutades exponential backoff retry loogikat ebaõnnestunud päringute korral.
    """
    all_data = []
    start = 0
    
    while True:
        end = start + batch_size - 1
        retries = 0
        success = False
        response_data = None
        
        # Retry loogika exponential backoff-iga
        while retries < max_retries and not success:
            try:
                # Teeme vahemikupõhise päringu (.range)
                response = supabase.table(table_name).select("*").range(start, end).execute()
                response_data = response.data
                success = True
            except Exception as e:
                retries += 1
                wait_time = 2 ** retries  # 2, 4, 8 sekundit ooteaega
                print(f"[HOIATUS] Katse {retries}/{max_retries} tabelile '{table_name}' ebaõnnestus: {e}")
                if retries < max_retries:
                    print(f"Ootan {wait_time} sekundit enne uuesti proovimist...")
                    time.sleep(wait_time)
                else:
                    print(f"[VIGA] Kõik katsed tabeli '{table_name}' pärimiseks nurjusid.")
                    raise e
                    
        # Kui andmeid ei tule enamat tagasi, lõpetame tsükli
        if not response_data:
            break
            
        all_data.extend(response_data)
        
        # Kui saime vähem ridu kui batch_size, tähendab see, et jõudsime tabeli lõppu
        if len(response_data) < batch_size:
            break
            
        start += batch_size
        
    return pd.DataFrame(all_data)

print("Pagination ja retry funktsioon on edukalt loodud!")
```

```python
# Roll B - 1. samm: Andmete puhastamise funktsioon
def clean_data(df: pd.DataFrame) -> pd.DataFrame:
    """Eemaldab duplikaadid, tegeleb puuduvate väärtustega ja teisendab kuupäeva."""
    if df.empty:
        return df
        
    df_clean = df.copy()
    
    # Eemalda duplikaadid invoice_id järgi
    if 'invoice_id' in df_clean.columns:
        df_clean = df_clean.drop_duplicates(subset=['invoice_id'])
    else:
        df_clean = df_clean.drop_duplicates()
        
    # Eemalda read, kus kriitilised väljad on NULL-id
    critical_cols = [c for c in ['customer_id', 'sale_date', 'total_price'] if c in df_clean.columns]
    if critical_cols:
        df_clean = df_clean.dropna(subset=critical_cols)
        
    # Teisenda kuupäev datetime formaati
    if 'sale_date' in df_clean.columns:
        df_clean['sale_date'] = pd.to_datetime(df_clean['sale_date'])
        
    # Eemalda negatiivsed hinnad
    if 'total_price' in df_clean.columns:
        df_clean = df_clean[df_clean['total_price'] > 0]
        
    return df_clean

print("Funktsioon clean_data() on edukalt loodud!")
```

```python
# Roll B - 2. samm: Andmestike liitmise funktsioon
def merge_datasets(df_sales: pd.DataFrame, df_customers: pd.DataFrame) -> pd.DataFrame:
    """Liidab müügi- ja kliendiandmed customer_id järgi (left join)."""
    if df_sales.empty:
        return df_sales
    if df_customers.empty:
        return df_sales
        
    # Kasutame left join-i, et säilitada kõik müügitehingud
    merged_df = pd.merge(df_sales, df_customers, on='customer_id', how='left')
    return merged_df

print("Funktsioon merge_datasets() on edukalt loodud!")
```

```python
# Roll B - 3. samm: Nädalaste koondnäitajate arvutamise funktsioon
def calculate_weekly_aggregates(df: pd.DataFrame) -> pd.DataFrame:
    """Grupeerib andmed nädalate kaupa ja arvutab tulu, tellimuste arvu ning keskmise."""
    if df.empty or 'sale_date' not in df.columns or 'total_price' not in df.columns:
        return pd.DataFrame()
        
    # Resample nädala alusel ('W' - viitab nädala lõpule/algusele)
    weekly = df.resample('W', on='sale_date').agg(
        total_revenue=('total_price', 'sum'),
        order_count=('sale_id', 'count') if 'sale_id' in df.columns else ('total_price', 'count'),
        avg_order_value=('total_price', 'mean')
    ).reset_index()
    
    return weekly

print("Funktsioon calculate_weekly_aggregates() on edukalt loodud!")
```

```python
# Roll B - 4. samm: KPI-de arvutamise funktsioon
def calculate_kpis(df: pd.DataFrame) -> dict:
    """Tagastab sõnastiku peamiste KPI-dega: kogutulu, unikaalsed kliendid, keskmine tellimuse väärtus."""
    if df.empty:
        return {"total_revenue": 0, "unique_customers": 0, "avg_order_value": 0}
        
    total_revenue = float(df['total_price'].sum()) if 'total_price' in df.columns else 0.0
    unique_customers = int(df['customer_id'].nunique()) if 'customer_id' in df.columns else 0
    avg_order_value = float(df['total_price'].mean()) if 'total_price' in df.columns else 0.0
    
    kpis = {
        "total_revenue": round(total_revenue, 2),
        "unique_customers": unique_customers,
        "avg_order_value": round(avg_order_value, 2)
    }
    
    return kpis

print("Funktsioon calculate_kpis() on edukalt loodud!")
```

```python
# Edasijõudnute tase: clean_data funktsioon valideerimise ja logimisega
import logging

# Seadistame lihtsa logimise
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')

def clean_data_advanced(df: pd.DataFrame) -> pd.DataFrame:
    """Puhastab andmed, valideerib tüübid ja vahemikud ning logib tegevused."""
    if df.empty:
        logging.warning("Sisend-DataFrame on tühi!")
        return df
        
    logging.info(f"Alustan andmete puhastamist. Algne ridade arv: {len(df)}")
    df_clean = df.copy()
    
    # 1. Duplikaatide eemaldamine
    before_dup = len(df_clean)
    if 'invoice_id' in df_clean.columns:
        df_clean = df_clean.drop_duplicates(subset=['invoice_id'])
    else:
        df_clean = df_clean.drop_duplicates()
    logging.info(f"Eemaldatud duplikaadid: {before_dup - len(df_clean)} rida.")
    
    # 2. Kriitiliste NULL-ide eemaldamine
    critical_cols = [c for c in ['customer_id', 'sale_date', 'total_price'] if c in df_clean.columns]
    before_na = len(df_clean)
    df_clean = df_clean.dropna(subset=critical_cols)
    logging.info(f"Eemaldatud puuduvate väärtustega read: {before_na - len(df_clean)} rida.")
    
    # 3. Andmetüübi ja vahemiku valideerimine (Edasijõudnute nõue)
    if 'total_price' in df_clean.columns:
        # Kontrollime, kas tüüp on numbriline
        if not pd.api.types.is_numeric_dtype(df_clean['total_price']):
            df_clean['total_price'] = pd.to_numeric(df_clean['total_price'], errors='coerce')
            
        # Valideerime vahemiku: hind peab olema > 0 ja olemuselt mõistlik (nt < 100 000)
        invalid_prices = (df_clean['total_price'] <= 0) | (df_clean['total_price'] > 100000)
        invalid_count = invalid_prices.sum()
        if invalid_count > 0:
            logging.warning(f"Leiti {invalid_count} ebamõistlikku või negatiivset total_price väärtust, eemaldan need.")
            df_clean = df_clean[~invalid_prices]
            
    # 4. Kuupäeva formaadi teisendamine ja valideerimine
    if 'sale_date' in df_clean.columns:
        df_clean['sale_date'] = pd.to_datetime(df_clean['sale_date'], errors='coerce')
        # Kontrollime, et kuupäevad oleksid mõistlikus vahemikus (nt alates 2020. aastast)
        df_clean = df_clean[df_clean['sale_date'] >= '2020-01-01']
        
    logging.info(f"Puhastamine edukalt lõpetatud. Lõplik ridade arv: {len(df_clean)}")
    return df_clean

print("Edasijõudnute clean_data_advanced() on loodud!")
```
